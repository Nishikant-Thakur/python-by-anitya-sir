# Practical 10 – Visualization using Seaborn
**Dataset:** Iris  
**Objective:** Visualize the Iris dataset using scatter, density, violin, heatmap, joint, and pair plots, and interpret each visualization.

The Iris dataset contains 150 observations of three species—Setosa, Versicolor, and Virginica—with four numerical features: sepal length, sepal width, petal length, and petal width.

## 1. Import Libraries and Load Dataset

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Load the Iris dataset
df = sns.load_dataset("iris")

# Display the first five rows
df.head()
```

## 2. Scatter Plot

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="sepal_length", y="sepal_width", hue="species", s=70)
plt.title("Scatter Plot: Sepal Length vs Sepal Width")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.show()
```

### Analysis
From the graph, Setosa usually has a shorter sepal length than the other two flowers. Versicolor and Virginica have longer sepals, and their points overlap in some places. This means that sepal length and sepal width are not enough to clearly identify every species. The scatter plot makes it easy to see the relationship between these two measurements.

## 3. Density Plot

```python
plt.figure(figsize=(8, 5))
sns.kdeplot(data=df, x="petal_length", hue="species", fill=True, common_norm=False, alpha=0.35)
plt.title("Density Plot: Petal Length by Species")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Density")
plt.show()
```

### Analysis
The density plot shows that Setosa has smaller petal lengths. Versicolor mostly falls in the middle range, while Virginica generally has longer petals. There is a little overlap between Versicolor and Virginica, but petal length still helps us tell the species apart.

## 4. Violin Plot

```python
plt.figure(figsize=(8, 5))
sns.violinplot(data=df, x="species", y="petal_length", inner="box")
plt.title("Violin Plot: Petal Length by Species")
plt.xlabel("Species")
plt.ylabel("Petal Length (cm)")
plt.show()
```

### Analysis
In this violin plot, Setosa has the shortest petals, Versicolor has medium-sized petals, and Virginica has the longest petals. The wider parts of each violin show where more values are grouped together. Overall, the plot gives a clear comparison of petal length and its spread for the three species.

## 5. Heatmap (Correlation Analysis)

```python
# Calculate correlation among numerical features only
correlation = df.drop(columns="species").corr()

plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f", vmin=-1, vmax=1)
plt.title("Correlation Heatmap of Iris Features")
plt.show()

correlation
```

### Analysis (Correlation)
The heatmap shows how the numerical features are related to each other. Petal length and petal width have a very strong positive correlation of about **0.96**, so both measurements usually increase together. Sepal length and petal length also have a strong positive correlation of around **0.87**. Sepal width and petal width have a negative correlation of about **−0.37**, while sepal length and sepal width have a weak negative correlation of about **−0.12**.

Correlation values range from −1 to +1. A positive value means the two features tend to increase together, while a negative value means one tends to decrease as the other increases. These values show a relationship, not that one feature causes the other.

## 6. Joint Plot

```python
sns.jointplot(data=df, x="petal_length", y="petal_width", hue="species", height=7)
plt.show()
```

### Analysis
The joint plot shows that petal length and petal width are closely related. In general, flowers with longer petals also have wider petals. The small graphs along the sides show the distribution of each measurement separately. Setosa forms a separate group, while Versicolor and Virginica have some overlap.

## 7. Pair Plot

```python
sns.pairplot(df, hue="species", diag_kind="hist")
plt.suptitle("Pair Plot of Iris Dataset", y=1.02)
plt.show()
```

### Analysis
The pair plot shows the relationship between all the numerical features in the Iris dataset. The graphs on the diagonal show the distribution of each feature. Petal length and petal width separate the species most clearly, especially Setosa. Versicolor and Virginica still overlap in some measurements, mainly in the sepal graphs. This plot is helpful for understanding the dataset as a whole.

From the graphs, I found that petal length and petal width are the most useful features for identifying the species. They also have a strong positive correlation, which means they usually increase together. Overall, these visualizations made it easier to understand the data and compare the three Iris species.
